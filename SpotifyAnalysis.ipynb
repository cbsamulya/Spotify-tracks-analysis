```python
genre_popularity = (
    df.groupby('track_genre')['popularity']
    .mean()
    .sort_values(ascending = False)
)
genre_popularity.head(10)
```

```python
genre_counts = df["track_genre"].value_counts()
genre_counts.head(10)
```

```python
top_genres = genre_popularity.head(10).index

df[df['track_genre'].isin(top_genres)].groupby('track_genre')[
    ['popularity', 'danceability', 'energy', 'valence', 'acousticness']
].mean().sort_values('popularity', ascending=False)
```

```python
df[['popularity', 'danceability', 'energy', 'valence',
    'acousticness', 'instrumentalness', 'speechiness',
    'liveness', 'tempo']].corr()['popularity'].sort_values(ascending=False)
```

```python
mood_genres = ['happy' , 'sad' , 'chill' , 'sleep' ,'study' , 'party']
```

```python
df[df['track_genre' ].isin(mood_genres)].shape
```

```python
mood_analysis = ( 
    df[df['track_genre'].isin(mood_genres)]
    .groupby('track_genre')[['energy' , 'valence' , 'danceability' , 'acousticness']]
    .mean()
)

mood_analysis
```
